```python
import pandas as pd
import requests
from io import StringIO 
url = 'https://en.wikipedia.org/wiki/List_of_largest_companies_in_the_United_States_by_revenue'
headers = {'User-Agent': 'Mozilla/5.0'}
response = requests.get(url, headers=headers)
html_io = StringIO(response.text)
tables = pd.read_html(html_io, flavor='html5lib')
df = tables[0]
print(df.head())
```

```python
df = df.head(10)
df
```

```python
# Display basic dataset information
df.info()

# Display statistical summary
df.describe()
```

```python
print(df.columns)
```

```python
# Create a horizontal bar chart showing revenue by company

plt.figure(figsize=(10,6))

sns.barplot(x="Revenue (USD millions)",        # X-axis represents revenue
            y="Name",           # Y-axis represents company names
            data=df,
            palette="magma")   # Color palette for dark theme

plt.title("Top 10 U.S. Companies by Revenue", fontsize=14)
plt.xlabel("Revenue (USD Millions)")
plt.ylabel("Company")

plt.show()
```

```python
print(df.columns.tolist())
```

```python
# Line chart to visualize how revenue changes with company rank

plt.figure(figsize=(8,5))

plt.plot(df["Rank"],            # Rank on X-axis
         df["Revenue (USD millions)"],         # Revenue on Y-axis
         marker="o",            # Circular markers
         linewidth=3,
         color="cyan")

# Invert X-axis so rank 1 appears first
plt.gca().invert_xaxis()

plt.title("Revenue (USD millions)")
plt.xlabel("Rank")
plt.ylabel("Revenue (USD Millions)")

plt.show()
```

```python
# Count how many companies belong to each industry
industry_counts = df["Industry"].value_counts()

plt.figure(figsize=(7,7))

plt.pie(industry_counts,
        labels=industry_counts.index,  # Industry labels
        autopct="%1.1f%%",             # Percentage values
        colors=sns.color_palette("bright"))

plt.title("Industry Distribution of Top Companies")

plt.show()
```

```python
print(df.columns.tolist())
```

```python
# Scatter plot to examine relationship between employees and revenue

plt.figure(figsize=(8,6))

sns.scatterplot(x="Employees",        # X-axis = number of employees
                y="Revenue (USD millions)",          # Y-axis = revenue
                data=df,
                hue="Industry",       # Color based on industry
                size="Revenue (USD millions)",       # Bubble size represents revenue
                sizes=(100,1000),
                palette="cool")

plt.title("Employees vs Revenue Relationship")

plt.show()
```

```python
# Bar chart to compare number of employees across companies

plt.figure(figsize=(10,5))

sns.barplot(x="Employees",
            y="Name",
            data=df,
            palette="viridis")

plt.title("Top Companies by Number of Employees")
plt.xlabel("Employees")
plt.ylabel("Company")

plt.show()
```

```python
# Create correlation matrix between numeric columns

corr = df[["Revenue (USD millions)","Employees"]].corr()

plt.figure(figsize=(5,4))

sns.heatmap(corr,
            annot=True,
            cmap="coolwarm")

plt.title("Correlation Between Revenue and Employees")

plt.show()
```

Walmart and Amazon currently lead the list in terms of revenue generation. A key observation is the frequent appearance of healthcare companies within the top ten, indicating their strong market position. Data analysis reveals a clear inverse relationship between company rank and revenue, where lower rankings generally correlate with decreased revenue. Furthermore, companies with larger workforces tend to demonstrate higher revenue figures. Collectively, the industry diversity among these top-tier organizations reflects the multifaceted nature of the economy.
